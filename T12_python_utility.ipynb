```python
def calculator():
    """Perform basic arithmetic operations."""

    try:
        first = float(input("Enter first number: "))
        second = float(input("Enter second number: "))

        print("1. Add")
        print("2. Subtract")
        print("3. Multiply")
        print("4. Divide")

        choice = input("Choose operation: ")

        if choice == "1":
            print("Result:", first + second)

        elif choice == "2":
            print("Result:", first - second)

        elif choice == "3":
            print("Result:", first * second)

        elif choice == "4":

            if second == 0:
                print("Cannot divide by zero.")
            else:
                print("Result:", first / second)

        else:
            print("Invalid operation.")

    except ValueError:
        print("Enter valid numbers.")


def palindrome_checker():
    """Check whether text is a palindrome."""

    text = input("Enter text: ").lower()

    reversed_text = ""

    for character in text:
        reversed_text = character + reversed_text

    if text == reversed_text:
        print("Palindrome")
    else:
        print("Not a palindrome")


def prime_checker():
    """Check whether number is prime."""

    number = int(input("Enter number: "))

    if number <= 1:
        print("Not prime")
        return

    for divisor in range(2, number):

        if number % divisor == 0:
            print("Not prime")
            return

    print("Prime number")


def factorial_calculator():
    """Calculate factorial."""

    number = int(input("Enter number: "))

    if number < 0:
        print("Factorial is not defined for negative numbers.")
        return

    factorial = 1

    for value in range(1, number + 1):
        factorial *= value

    print("Factorial:", factorial)


def multiplication_table():
    """Display multiplication table."""

    number = int(input("Enter number: "))

    for value in range(1, 21):

        print(
            number,
            "x",
            value,
            "=",
            number * value
        )


def password_checker():
    """Perform basic password strength check."""

    password = input("Enter password: ")

    uppercase = False
    lowercase = False
    number = False

    for character in password:

        if character.isupper():
            uppercase = True

        if character.islower():
            lowercase = True

        if character.isdigit():
            number = True

    if (
        len(password) >= 8
        and uppercase
        and lowercase
        and number
    ):
        print("Strong password")
    else:
        print("Weak password")


while True:

    print("\n==============================")
    print(" SUPER30 PYTHON UTILITY APP")
    print("==============================")

    print("1. Calculator")
    print("2. Palindrome Checker")
    print("3. Prime Checker")
    print("4. Factorial Calculator")
    print("5. Multiplication Table")
    print("6. Password Checker")
    print("7. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        calculator()

    elif choice == "2":
        palindrome_checker()

    elif choice == "3":
        prime_checker()

    elif choice == "4":
        factorial_calculator()

    elif choice == "5":
        multiplication_table()

    elif choice == "6":
        password_checker()

    elif choice == "7":
        print("Thank you for using Super30 Python Utility App!")
        break

    else:
        print("Invalid choice. Please try again.")
```
