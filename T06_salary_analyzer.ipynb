```python
def total_payroll(salaries):
    """Return total employee payroll."""
    total = 0

    for salary in salaries:
        total = total + salary

    return total


def average_salary(salaries):
    """Return average salary."""
    return total_payroll(salaries) / len(salaries)


def highest_salary(salaries):
    """Find highest salary without max()."""
    highest = salaries[0]

    for salary in salaries:
        if salary > highest:
            highest = salary

    return highest


def lowest_salary(salaries):
    """Find lowest salary without min()."""
    lowest = salaries[0]

    for salary in salaries:
        if salary < lowest:
            lowest = salary

    return lowest


def above_average(salaries):
    """Return salaries greater than average."""
    average = average_salary(salaries)

    result = []

    for salary in salaries:
        if salary > average:
            result.append(salary)

    return result


salaries = [45000, 60000, 75000, 50000, 90000]

print("Total payroll:", total_payroll(salaries))
print("Average:", average_salary(salaries))
print("Highest:", highest_salary(salaries))
print("Lowest:", lowest_salary(salaries))
print("Above average:", above_average(salaries))
```
