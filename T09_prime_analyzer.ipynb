```python
def is_prime(number):
    """Return True if number is prime."""

    if number <= 1:
        return False

    for divisor in range(2, number):

        if number % divisor == 0:
            return False

    return True


def find_primes(start, end):
    """Return prime numbers within a range."""

    primes = []

    for number in range(start, end + 1):

        if is_prime(number):
            primes.append(number)

    return primes


def prime_sum(primes):
    """Calculate sum of prime numbers."""

    total = 0

    for prime in primes:
        total = total + prime

    return total


def largest_prime(primes):
    """Find largest prime."""

    if len(primes) == 0:
        return None

    largest = primes[0]

    for prime in primes:

        if prime > largest:
            largest = prime

    return largest


start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))
print(f"Finding prime numbers between {start} and {end}")
primes = find_primes(start, end)

print("Prime numbers:", primes)
print("Prime count:", len(primes))
print("Prime sum:", prime_sum(primes))
print("Largest prime:", largest_prime(primes))
```
