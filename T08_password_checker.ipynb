```python
def check_password(password):
    """Check password strength based on common requirements."""

    has_uppercase = False
    has_lowercase = False
    has_number = False
    has_special = False

    special_characters = "!@#$%^&*()_+-="

    for character in password:

        if character.isupper():
            has_uppercase = True

        elif character.islower():
            has_lowercase = True

        elif character.isdigit():
            has_number = True

        elif character in special_characters:
            has_special = True

    has_minimum_length = len(password) >= 8

    if (
        has_uppercase
        and has_lowercase
        and has_number
        and has_special
        and has_minimum_length
    ):
        return "Strong Password"

    print("\nPassword requirements:")

    if not has_minimum_length:
        print("- Minimum 8 characters")

    if not has_uppercase:
        print("- Add uppercase letter")

    if not has_lowercase:
        print("- Add lowercase letter")

    if not has_number:
        print("- Add number")

    if not has_special:
        print("- Add special character")

    return "Weak Password"


password = input("Enter password: ")

print(check_password(password))
```
