```python
cart = []


def add_product():
    """Add product to cart."""
    name = input("Enter product name: ")

    try:
        price = float(input("Enter price: ₹"))

        cart.append({
            "name": name,
            "price": price
        })

        print("Product added.")

    except ValueError:
        print("Invalid price.")


def remove_product():
    """Remove product from cart."""
    name = input("Enter product to remove: ")

    for product in cart:

        if product["name"].lower() == name.lower():
            cart.remove(product)
            print("Product removed.")
            return

    print("Product not found.")


def view_cart():
    """Display shopping cart."""
    if len(cart) == 0:
        print("Cart is empty.")
        return

    for product in cart:
        print(product["name"], "- ₹", product["price"])


def calculate_bill():
    """Calculate cart total."""
    total = 0

    for product in cart:
        total = total + product["price"]

    print("Total bill: ₹", total)


while True:

    print("\n--- SHOPPING CART ---")
    print("1. Add Product")
    print("2. Remove Product")
    print("3. View Cart")
    print("4. Calculate Bill")
    print("5. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        add_product()

    elif choice == "2":
        remove_product()

    elif choice == "3":
        view_cart()

    elif choice == "4":
        calculate_bill()

    elif choice == "5":
        break

    else:
        print("Invalid choice.")
```
